# Multi-Asset Portfolio Risk & Performance Exploratory Analysis

This notebook demonstrates exploratory quantitative analysis, risk modeling, Markowitz optimization, and Monte Carlo simulation using the Data & Analytics Engine.

```python
import sys
import os
sys.path.insert(0, os.path.abspath('../../src'))

from portfolio_analytics.database.connection import get_db_session
from portfolio_analytics.pipelines.analytics_pipeline import ComprehensiveAnalyticsPipeline

# Execute analytics report extraction
with get_db_session() as db:
    pipeline = ComprehensiveAnalyticsPipeline(db)
    report = pipeline.generate_report(portfolio_id=1)

print(f"Portfolio: {report.snapshot.name}")
print(f"Market Value: ${report.snapshot.total_market_value:,.2f}")
print(f"Unrealized PnL: ${report.snapshot.total_unrealized_pnl:,.2f} ({report.snapshot.total_unrealized_pnl_pct:+.2f}%)")
print(f"Annualized Volatility: {report.risk.annualized_volatility * 100:.2f}%")
print(f"Sharpe Ratio: {report.risk.sharpe_ratio:.3f}")
print(f"Sortino Ratio: {report.risk.sortino_ratio:.3f}")
print(f"Max Drawdown: {report.risk.max_drawdown * 100:.2f}%")
print(f"Historical VaR (95%): {report.risk.historical_var_95 * 100:.2f}%")
```

## Euler Risk Attribution Analysis
Decomposing total portfolio volatility into percentage contributions to risk (PCR) per holding.

```python
for item in report.risk_attribution.asset_contributions:
    print(f"{item.symbol:6s} | Weight: {item.weight * 100:5.2f}% | Risk Contrib: {item.percentage_contribution * 100:5.2f}%")
print(f"Sum of Risk Contributions: {report.risk_attribution.sum_percentage_contributions * 100:.2f}%")
```

## Macroeconomic Stress Testing
Evaluating portfolio market value change under major historical market dislocations.

```python
for test in report.stress_tests:
    print(f"{test.scenario_name:32s} | PnL: ${test.impact_dollar:+10.2f} ({test.impact_pct:+6.2f}%)")
```
